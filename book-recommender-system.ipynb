POPULARITY BASED RECOMMENDER SYSTEM CONSIDERING BOOKS WITH HIGHEST RATING WITH MNIMUM 250 RATINGS

```python
ratings_with_name=ratings.merge(books,on='ISBN')
```

```python
num_rating_df=ratings_with_name.groupby('Book-Title').count()['Book-Rating'].reset_index()
num_rating_df.rename(columns={'Book-Rating':'num_ratings'},inplace=True)
num_rating_df
```

```python
ratings_with_name['Book-Rating'].dtype
```

```python
avg_rating_df=ratings_with_name.groupby('Book-Title')['Book-Rating'].mean().reset_index(name='avg_ratings')
avg_rating_df.rename(columns={'Book-Rating':'avg_ratings'},inplace=True)
avg_rating_df
```

```python
ratings_with_name['Book-Rating'].dtype
```

```python
popular_df=num_rating_df.merge(avg_rating_df,on='Book-Title')
popular_df
```

```python
popular_df=popular_df[popular_df['num_ratings']>=250].sort_values('avg_ratings',ascending=False).head(50)
```

```python
popular_df=popular_df.merge(books,on='Book-Title').drop_duplicates('Book-Title')[['Book-Title','Book-Author','Image-URL-M','num_ratings','avg_ratings']]
```

COLLABORATIVE FILTERING BASED RECOMMENDATION SYSTEM , will take users with 200 and greater ratings on books and considering the books which have min 50 people rating

```python
x=ratings_with_name.groupby('User-ID').count()['Book-Rating']>200
bibliophile=x[x].index
```

```python
filtered_rating=ratings_with_name = ratings_with_name[
    ratings_with_name['User-ID'].isin(bibliophile)
]
filtered_rating
```

```python
y=filtered_rating.groupby('Book-Title').count()['Book-Rating']>=50
famous_books=y[y].index
```

```python
final_ratings=filtered_rating[filtered_rating['Book-Title'].isin(famous_books)]
```

```python
pt=final_ratings.pivot_table(index='Book-Title',columns='User-ID',values='Book-Rating')
```

```python
pt.fillna(0,inplace=True)
```

```python
from sklearn.metrics.pairwise import cosine_similarity 
```

```python
pt = pt.head(2000)
```

Similarity score of every book wrt to each other book in the pt table

```python
pt = pt.head(2000)
similarity_score=cosine_similarity(pt)
similarity_score
```

```python
print("PT shape:", pt.shape)
```

```python
cosine_similarity(pt).shape
```

```python
import random
import numpy as np

# AFTER loading pt pickle file
pt_index_lookup = {title.lower(): title for title in pt.index}


def recommend(book_name):

    # Normalize user input
    book_name = book_name.strip().lower()

    # Case-insensitive lookup
    if book_name not in pt_index_lookup:

        data = []

        random_books = popular_df.sample(5)

        for _, row in random_books.iterrows():
            item = [
                row['Book-Title'],
                row['Book-Author'],
                row['Image-URL-M']
            ]
            data.append(item)

        return data

    # Get correct title
    actual_title = pt_index_lookup[book_name]

    index = np.where(pt.index == actual_title)[0][0]
    distances = similarity_score[index]

    similar_items = sorted(
        list(enumerate(distances)),
        key=lambda x: x[1],
        reverse=True
    )[1:6]

    data = []

    for i in similar_items:
        temp_df = books[books['Book-Title'] == pt.index[i[0]]]

        item = [
            temp_df['Book-Title'].values[0],
            temp_df['Book-Author'].values[0],
            temp_df['Image-URL-M'].values[0]
        ]

        data.append(item)

    return data
```

```python
recommend('harry potter')
```
